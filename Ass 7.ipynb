# Part A — Understanding the HPC Environment

## A1. What is the purpose of a login node?

The login node is the entry point to the HPC system. Users connect to it to access the HPC environment, manage files, create and edit scripts, submit jobs to SLURM, and monitor their jobs.

In this HPC environment, the login node is identified as `betinserver`. The login node is mainly used for access and job management rather than computationally intensive analyses.

## A2. Why should computationally intensive analyses not normally be run directly on the login node?

Computationally intensive analyses should not normally be run directly on the login node because the login node is a shared resource used by many users to access and manage the HPC system.

Running heavy bioinformatics analyses directly on the login node can consume substantial CPU, memory, and input/output resources and may interfere with other users.

Instead, intensive analyses should be submitted through SLURM so that they can be scheduled using the appropriate computational resources.

## A3. What is the purpose of a compute node?

A compute node is a machine that performs the computational work requested by a user's job. It provides resources such as CPU cores and memory for running bioinformatics programs and other computationally intensive analyses.

Jobs submitted through SLURM are scheduled to run on available compute resources.

## A4. What role does SLURM play between the user and the compute nodes?

SLURM acts as the workload manager and job scheduler between the user and the compute resources.

The user submits a job and specifies the resources required, such as:

- CPUs
- Memory
- Wall-clock time

SLURM manages the job queue and schedules jobs according to available resources and the cluster configuration. It also allows users to monitor and manage their jobs.

## A5. HPC architecture

The basic HPC architecture can be represented as:

```text
                         USER
                           │
                           │ SSH
                           ▼
                  ┌──────────────────┐
                  │    LOGIN NODE    │
                  │    betinserver   │
                  └────────┬─────────┘
                           │
                           │ Submit job
                           ▼
                  ┌──────────────────┐
                  │      SLURM       │
                  │  Job Scheduler   │
                  └────────┬─────────┘
                           │
                           │ Allocate resources
                           ▼
                  ┌──────────────────┐
                  │   COMPUTE NODE   │
                  │  Run analysis    │
                  └────────┬─────────┘
                           │
                           ▼
                        RESULTS
```

The user first connects to the HPC through the login node using SSH. The user prepares files and scripts on the login node and submits computational jobs to SLURM.

SLURM manages the job queue and allocates the requested computational resources. The job then executes on an appropriate compute node, and the resulting output files are stored in the user's designated directories.

## Why is separating `data/`, `scripts/`, `logs/`, and `results/` useful for reproducible bioinformatics?

Separating project files into different directories keeps a bioinformatics project organized and makes the analysis easier to reproduce and manage.

### `data/`

The `data/` directory stores input data used for the analysis, such as:

- FASTQ files
- FASTA files
- BAM files
- Other biological input data

The `scripts/` directory stores commands and scripts used to perform the analysis.

- Job execution information

Separating results from original data helps prevent accidental modification of input files.

### `tmp/`

The `tmp/` directory can be used for temporary or intermediate files that do not need to be retained as final results.

This organization improves reproducibility because the input data, analysis procedures, execution records, and results are kept separate and clearly identified.

## SLURM job states

### PENDING

The job has been submitted but has not started running. It may be waiting for resources or another scheduling condition.

### RUNNING

The job has started and is currently executing.

Job 3 was observed in this state.

### COMPLETED

The job finished successfully.

### FAILED

The job terminated unsuccessfully, usually because the program or script encountered an error.

### CANCELLED

The job was stopped before normal completion, for example by the user or system administrator.

### TIMEOUT

The job exceeded its allocated wall-clock time limit before finishing.

### OUT_OF_MEMORY

The job exceeded the memory allocated to it and was terminated because it used more memory than requested.

## Wall time versus CPU time

**Wall time** is the actual elapsed time between the start and end of a job. It is similar to measuring a job using a stopwatch.

**CPU time** is the amount of processor time used by the program.

CPU time can differ from wall time because a program may spend time:

- Waiting for input/output
- Sleeping
- Waiting for other processes
- Performing calculations

For example, the monitoring script contained:

```bash
sleep 180
```

During the sleep period, the job continues to consume wall-clock time but does not continuously perform CPU calculations.

Therefore:

- **Wall time:** How long the job takes in real-world time.
- **CPU time:** How much processor time the program consumes.

## F1. What happens if 32 CPUs are requested but the program uses only 1 thread?

The job would be inefficient because SLURM reserves 32 CPUs while the program uses only one CPU thread.

Most of the allocated CPU resources would remain unused. This can reduce overall cluster efficiency and may also increase the amount of time the job waits for resources.

## F2. What could happen if SLURM allocates 4 CPUs but the program launches 32 threads?

This creates **CPU oversubscription**.

Thirty-two threads would compete for only four allocated CPUs. The program may experience reduced performance because many threads must share the available CPU resources.

It also creates a mismatch between the resources allocated by SLURM and the resources requested by the application.

## F3. Why is `program --threads "$SLURM_CPUS_PER_TASK"` preferred over `program --threads 32`?

Using:

```bash
program --threads "$SLURM_CPUS_PER_TASK"
```

automatically makes the number of application threads match the number of CPUs allocated by SLURM.

For example, if the job is changed from 8 CPUs to 16 CPUs, the program automatically uses 16 threads without requiring the command to be manually changed.

Hardcoding:

```bash
program --threads 32
```

can cause a mismatch. The program might use more threads than the allocated CPUs, or allocated CPUs might remain unused.

Using `$SLURM_CPUS_PER_TASK` therefore makes the job more flexible, efficient, and consistent with the scheduler allocation.

# Part H — Controlling Array Concurrency

Suppose there are 1,000 samples.

Submitting 1,000 tasks simultaneously could place excessive demand on shared HPC resources.

The array can therefore be limited using:

```bash
#SBATCH --array=1-1000%20
```

The `1-1000` specifies 1,000 array tasks.

The `%20` specifies that a maximum of 20 tasks may run simultaneously.

## CPU calculation

Each running task uses 8 CPUs.

At most 20 tasks can run simultaneously:

```text
20 × 8 = 160 CPUs
```

Therefore:

**Maximum simultaneous CPU allocation = 160 CPUs**

## RAM calculation

Each running task uses 24 GB RAM.

At most 20 tasks can run simultaneously:

```text
20 × 24 GB = 480 GB
```

Therefore:

**Maximum simultaneous memory allocation = 480 GB RAM**

## Why is limiting concurrency beneficial?

Limiting array concurrency prevents all 1,000 tasks from running simultaneously and potentially overwhelming shared HPC resources.

It controls:

- CPU demand
- Memory demand
- Storage/I/O pressure
- Scheduler load

It also improves fairness for other users.

In this example, only 20 tasks can run at once. As tasks finish, SLURM can start additional tasks until all 1,000 samples have been processed.

# Part I — Job Dependencies

The bioinformatics workflow consists of several sequential stages:

```text
FASTQ
  ↓
Alignment
  ↓
BAM sorting
  ↓
Variant calling
```

Variant calling should not begin until the required preceding processing stages have completed successfully.

If the alignment job has Job ID `12345`, the next job can be submitted using:

```bash
sbatch --dependency=afterok:12345 variant_calling.sh
```

The `afterok` dependency tells SLURM to run `variant_calling.sh` only after Job `12345` has completed successfully.

## Why are dependencies preferable to manually checking whether every preceding job has finished?

SLURM job dependencies allow the scheduler to automatically control the order of jobs according to their completion status.

The user does not need to repeatedly check whether every preceding job has finished before submitting the next stage.

This is particularly useful for bioinformatics workflows containing many samples and multiple processing stages.

Dependencies:

- Reduce manual intervention
- Reduce human error
- Prevent downstream jobs from starting too early
- Allow workflows to run automatically
- Ensure downstream jobs depend on successful upstream jobs

For example:

```text
Alignment Job 12345
        │
        ▼
  Completes OK
        │
        ▼
Variant Calling Job
        │
        ▼
      Starts
```

If the alignment job fails, an `afterok` dependency prevents the downstream job from proceeding as though the required input had been successfully generated.

### Important workflow consideration

If BAM sorting is a separate SLURM job, variant calling should actually depend on the **sorting job**, because sorted BAM files are the required inputs.

For example:

```text
Alignment
    ↓
Sorting
    ↓
Variant Calling
```

The dependencies should therefore follow the actual workflow.

```bash
#SBATCH --mem=20G
```

The goal is to provide enough memory for successful completion while maintaining efficient resource usage.

## Scenario 2 — TIMEOUT

### Given

- State: `TIMEOUT`
- Requested time: `02:00:00`
- The log shows that the program was still processing when the scheduler terminated it.

Before simply increasing the wall time, investigate why the program did not finish within two hours.

The following should be checked:

1. **Examine the log files** to determine which step was still running.
2. **Check input data size** to determine whether the input was larger or more complex than expected.
3. **Check CPU utilization** to determine whether the program was using the allocated CPUs efficiently.
4. **Check memory usage** to identify possible memory pressure.
5. **Check the number of threads/CPUs** used by the program.
6. **Check for I/O bottlenecks**, such as slow disk access.
7. **Check program parameters** for inefficient settings.
8. Determine whether the analysis normally requires more than two hours or whether something unusual caused the slowdown.

After identifying the cause, an appropriate solution can be chosen.

If the program is working normally but genuinely requires more time, the wall-time request should be increased.

For example, if investigation shows that the job normally requires about three hours:

```bash
#SBATCH --time=03:30:00
```

If the problem is inefficient resource usage, however, the underlying problem should be fixed rather than simply requesting a much longer wall time.

## Scenario 3 — Excessive Resource Request

### Requested resources

```text
64 CPUs
256 GB RAM
48 hours
```

### Actual resource usage

```text
CPU utilization: very low
Peak RAM: 12 GB
Runtime: 35 minutes
```

This is poor HPC practice because the student requested far more resources than the job actually used.

The requested 256 GB of RAM is particularly excessive because the observed peak memory was only 12 GB.

Similarly, requesting 64 CPUs is inappropriate when CPU utilization was very low.

Finally, requesting 48 hours is excessive when the job completed in only 35 minutes.

Over-requesting resources can:

- Waste CPU resources
- Waste memory resources
- Make other users wait longer
- Reduce overall cluster efficiency
- Make the job harder for the scheduler to place
- Prevent other jobs from running unnecessarily

HPC users should therefore request resources based on measured or reasonably estimated requirements, with an appropriate safety margin.

### Example of a more appropriate request

An illustrative smaller request could be:

```bash
#SBATCH --cpus-per-task=4
#SBATCH --mem=16G
#SBATCH --time=01:00:00
```

This is an **illustrative right-sized request**, not an exact optimal request, because the scenario does not provide enough information to determine the ideal CPU count.

The reasoning is:

- **CPUs:** CPU utilization was very low, so substantially fewer than 64 CPUs should be requested. The exact number should be determined from measurements.
- **RAM:** Peak usage was 12 GB, so 256 GB is excessive. 16 GB provides a reasonable safety margin.
- **Time:** The job completed in 35 minutes, so 48 hours is excessive. One hour provides additional time while avoiding a very large unnecessary reservation.

The best practice is to monitor completed jobs and use actual resource usage to progressively **right-size future resource requests**.

## Overall lesson

HPC resource requests should be based on the program's actual requirements rather than simply requesting the largest possible resources.

A good request provides enough CPU, memory, and wall time for reliable completion while maintaining a reasonable safety margin and allowing the cluster to be shared efficiently.

# Part K — Final Bioinformatics Workflow Design

The final workflow is designed for paired-end RNA-seq FASTQ samples obtained by converting the SRA files from the previous assignment.

The goal is to process the samples through:

```text
FASTQ
  ↓
Quality Control
  ↓
Alignment
  ↓
Sorted BAM
  ↓
Quantification
  ↓
Expression Matrix
  ↓
Downstream Analysis
```

The workflow uses SLURM job arrays for independent sample-level processing and SLURM dependencies to control the order between workflow stages.

## K1. HPC strategy

A **SLURM job array** is appropriate because several stages can be performed independently for each biological sample.

These include:

- Quality control
- Alignment
- BAM sorting
- Quantification

The required resources for each alignment are approximately:

- **CPUs:** 8
- **RAM:** 24 GB
- **Wall time:** 90 minutes

Therefore, the alignment job should request:

```bash
#SBATCH --cpus-per-task=8
#SBATCH --mem=24G
#SBATCH --time=01:30:00
```

For example, if there are 100 samples, an array could be configured as:

```bash
#SBATCH --array=1-100%20
#SBATCH --cpus-per-task=8
#SBATCH --mem=24G
#SBATCH --time=01:30:00
```

The `1-100` specifies 100 sample-specific tasks, while `%20` limits the number of simultaneously running tasks to 20.

With 20 tasks running simultaneously:

```text
20 × 8 CPUs = 160 CPUs
```

and:

```text
20 × 24 GB = 480 GB RAM
```

The concurrency limit helps prevent the workflow from consuming excessive shared HPC resources.

## K5. Complete workflow

The complete RNA-seq workflow is:

```text
FASTQ
  ↓
Quality Control
  ↓
Alignment
  ↓
Sorted BAM
  ↓
Quantification
  ↓
Expression Matrix
  ↓
Downstream Analysis
```

For multiple samples, the sample-level stages can be parallelized:

```text
                         FASTQ
                           │
             ┌─────────────┼─────────────┐
             ▼             ▼             ▼
          Sample 1      Sample 2      Sample 3   ... Sample N
             │             │             │
             ▼             ▼             ▼
             QC            QC            QC
             │             │             │
             ▼             ▼             ▼
         Alignment     Alignment     Alignment
             │             │             │
             ▼             ▼             ▼
         Sorted BAM    Sorted BAM    Sorted BAM
             │             │             │
             ▼             ▼             ▼
         Quantification for each sample
             │             │             │
             └─────────────┼─────────────┘
                           ▼
                   Expression Matrix
                           │
                           ▼
                   Downstream Analysis
```

## K6. Stages that can be parallelized by sample

### Quality Control

Each sample's FASTQ files can be analyzed independently.

Therefore, QC is well suited to a job array.

### Alignment

Each sample can be aligned independently to the same reference genome.

This is an especially suitable stage for a SLURM job array.

Each alignment task requires approximately:

```text
8 CPUs
24 GB RAM
90 minutes
```

### BAM sorting

Each sample's alignment output can be sorted independently after its alignment has completed.

### Quantification

Sample-level quantification can generally be performed independently before the results are combined.

## K7. Stages requiring combined results

### Expression matrix construction

The individual sample-level quantification results must generally be available before they can be combined into an expression matrix.

Therefore, this stage should occur after the sample-level quantification jobs have completed successfully.

### Downstream analysis

Downstream analysis is performed using the combined expression matrix and therefore generally occurs after matrix construction.

## K8. Job dependencies

SLURM dependencies should be used to ensure that each stage begins only after its required upstream stage has completed successfully.

For example:

```bash
sbatch --dependency=afterok:<QC_JOB_ID> alignment.sh
```

followed by:

```bash
sbatch --dependency=afterok:<ALIGNMENT_JOB_ID> sorting.sh
```

and then:

```bash
sbatch --dependency=afterok:<SORTING_JOB_ID> quantification.sh
```

Finally, after all sample-level quantification jobs have successfully completed, the expression matrix can be generated.

The overall dependency structure is:

```text
QC
 ↓
Alignment
 ↓
Sorting
 ↓
Quantification
 ↓
Expression Matrix
 ↓
Downstream Analysis
```

Dependencies are preferable to manually checking job completion because SLURM automatically controls the workflow order and prevents downstream jobs from starting before successful completion of required upstream jobs.
